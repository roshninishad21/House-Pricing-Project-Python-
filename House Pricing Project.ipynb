```python
# Data Preprocessing
train_df=pd.get_dummies(train_df)
test_df=pd.get_dummies(test_df)
```

```python
# Align columns between train and test datasets
train_df,test_df=train_df.align(test_df,join='left',axis=1,fill_value=0)
```

```python
# Feature Selection (using Lasso for feature selection)
X_train=train_df.drop(columns=['SalePrice'])
y_train=train_df['SalePrice']
```

```python
X_test=test_df.drop(columns=['SalePrice'], errors='ignore')  # Test set does not have SalePrice
y_test=test_df['SalePrice'] if 'SalePrice' in test_df.columns else None
```

```python
# Features Scaling
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
X_train_scaled=ss.fit_transform(X_train)
X_test_scaled=ss.transform(X_test)
```

```python
# Lasso for feature selection
from sklearn.linear_model import Lasso
lasso=Lasso(alpha=0.1)  # Tune alpha as needed
lasso.fit(X_train_scaled, y_train)
```

```python
# Selected features
selected_features=X_train.columns[(lasso.coef_ != 0)]
print("Selected Features by Lasso:", selected_features)
```

```python
# Train-Test Split for training set
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val=train_test_split(X_train_scaled, y_train, test_size=0.2, random_state=42)
```

```python
# Build Lasso model with GridSearchCV
from sklearn.model_selection import GridSearchCV
param_grid={'alpha': [0.01, 0.1, 1, 10, 100]}
grid_search=GridSearchCV(estimator=Lasso(), param_grid=param_grid, cv=5, scoring='neg_mean_squared_error')
grid_search.fit(X_train, y_train)
```

```python
# Best parameters from GridSearchCV
print(f"Best alpha for Lasso: {grid_search.best_params_}")
```

```python
# Model Prediction
best_lasso=grid_search.best_estimator_
```

```python
# Train and validation predictions
y_train_pred = best_lasso.predict(X_train)
y_val_pred = best_lasso.predict(X_val)
```

```python
# Evaluate the model with MSE, RMSE, MAE, R2
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mse_train = mean_squared_error(y_train, y_train_pred)
rmse_train = np.sqrt(mse_train)
mae_train = mean_absolute_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)
```

```python
mse_val = mean_squared_error(y_val, y_val_pred)
rmse_val = np.sqrt(mse_val)
mae_val = mean_absolute_error(y_val, y_val_pred)
r2_val = r2_score(y_val, y_val_pred)
```

```python
# Print evaluation metrics
print(f"Training MSE: {mse_train}, RMSE: {rmse_train}, MAE: {mae_train}, R2: {r2_train}")
print(f"Validation MSE: {mse_val}, RMSE: {rmse_val}, MAE: {mae_val}, R2: {r2_val}")
```

```python
# Predict the SalePrice for the test set
y_test_pred_final=best_lasso.predict(X_test_scaled)
```

```python
# Prepare the final submission file
submission=pd.DataFrame({'Id': test_df['Id'], 'SalePrice': y_test_pred_final})
submission.to_csv('submission.csv', index=False)
print("Submission file created: submission.csv")
```
